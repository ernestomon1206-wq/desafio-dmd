```python
import pandas as pd


comerciantes = pd.read_csv("comerciantes_mercado.csv")
quiebra = pd.read_csv("quiebra_mercado.csv")
transacciones = pd.read_csv("transacciones_mercado.csv")

print("=== 1. COMERCIANTES MERCADO ===")
print(comerciantes.info())
print("\nValores Nulos:\n", comerciantes.isnull().sum())
print("\nEstadísticas:\n", comerciantes.describe(include='all'))

print("\n" + "="*40 + "\n=== 2. QUIEBRA MERCADO ===")
print(quiebra.info())
print("\nValores Nulos:\n", quiebra.isnull().sum())
print("\nEstadísticas:\n", quiebra.describe(include='all'))

print("\n" + "="*40 + "\n=== 3. TRANSACCIONES MERCADO ===")
print(transacciones.info())
print("\nValores Nulos:\n", transacciones.isnull().sum())
print("\nPrimeras filas:\n", transacciones.head())
```
